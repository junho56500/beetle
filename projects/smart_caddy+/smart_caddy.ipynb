Objectives:
- provide total service for player like tracking ball, driving cart autonomosly
- Safety first -> not driving ahead of all players, policy can be overtake slow teams
- manage players along with course status -> manage rounds effectively

Composition
- tracking ball : track ball from tee shot area (start point) by static installed camera( 2 cams behind area)
    - it can be extended to use cart cam to track ball after tee shot on later deployment
    - this model estimates trajectory of ball and ball drop area within 10 meters error detecting ball hitted on driving range
    - detect small and fast object

- ad cart
    - E2E autonomous driving model on golf cart following given destination from tracking ball application
    - input : destination and map and cam (front and end)
    - cam can be added to detect ball
    - this model estimates driving path and speed profile
    - control golf cart based on path and profile

- conditions
    - speed limit : 30kph
    - within golf course
    - managed adverse weather
    - managed map
    - not much objects (human, other carts, animal)
    - various geometry (slope, hill, narrow path,)
